## Charts as functions
Test-data prototypes below (columns from `DATA_CONTRACT.txt`). Each takes a DataFrame and returns a Plotly figure.
When moving one into `dashboard/charts.py`, keep the `(df, x_col, y_col)` signature the app expects, or update `app.py` to match.

```python
def sales_by_category(df):
    agg = df.groupby('category', as_index=False)['sales'].sum().sort_values('sales', ascending=False)
    return px.bar(agg, x='category', y='sales', title='Sales by category')


def sales_by_month(df):
    d = df.assign(month=df['order_date'].dt.to_period('M').astype(str))
    agg = d.groupby('month', as_index=False)['sales'].sum()
    return px.line(agg, x='month', y='sales', markers=True, title='Monthly sales')


sales_by_category(df).show()
sales_by_month(df).show()
```

## Numbers for the insights

```python
total = df['sales'].sum()
by_cat = df.groupby('category')['sales'].sum().sort_values(ascending=False)
by_month = df.groupby(df['order_date'].dt.to_period('M'))['sales'].sum()

print(f'Total sales: {total:,.2f}')
print(f'Top category: {by_cat.index[0]} = {by_cat.iloc[0] / total:.0%} of total')
print(f'Best month: {by_month.idxmax()} = {by_month.max():,.2f}')
```
